# Task
Fetch historical stock data for symbols AMAT, T, BIIB, CAT, CIM, CSCO, INTC, JCI, JPM, MDT, NVT, ORCL, QQM, QQQ, and JEPQ from the Alpha Vantage API, convert the data for each symbol into a pandas DataFrame, and save each DataFrame to a separate CSV file named after the symbol.

## Define symbols and api key

### Subtask:
Create a list of stock symbols and store your Alpha Vantage API key.


**Reasoning**:
Create a list of symbols and retrieve the API key as instructed.



```python
symbols = ["AMAT", "T", "BIIB", "CAT", "CIM", "CSCO", "INTC", "JCI", "JPM", "MDT", "NVT", "ORCL", "QQQM", "QQQ", "JEPQ"]
api_key = userdata.get('alphavantage')
```

## Fetch and process data

### Subtask:
Iterate through the list of symbols, fetch data from the Alpha Vantage API for each symbol, and convert the relevant part of the JSON response into a pandas DataFrame.


**Reasoning**:
Iterate through the symbols, fetch data for each, extract the time series, and create a DataFrame for each symbol, storing them in a dictionary.



```python
import pandas as pd

stock_dataframes = {}
base_url = 'https://www.alphavantage.co/query?function=TIME_SERIES_WEEKLY_ADJUSTED&symbol='

for symbol in symbols:
    url = f'{base_url}{symbol}&apikey={api_key}'
    r = requests.get(url)
    data = r.json()

    # Check if the expected key exists in the response
    if 'Weekly Adjusted Time Series' in data:
        time_series_data = data['Weekly Adjusted Time Series']
        df = pd.DataFrame.from_dict(time_series_data, orient='index')

        # Rename columns to remove the number prefix
        df.columns = [col.split('. ')[1] for col in df.columns]

        # Convert columns to numeric types, handling errors
        for col in df.columns:
            df[col] = pd.to_numeric(df[col], errors='coerce')

        stock_dataframes[symbol] = df
    else:
        print(f"Could not fetch data for symbol: {symbol}. Response: {data}")

# Display the first few rows of a sample DataFrame to verify
if stock_dataframes:
    sample_symbol = list(stock_dataframes.keys())[0]
    print(f"Sample DataFrame for {sample_symbol}:")
    display(stock_dataframes[sample_symbol].head())
else:
    print("No dataframes were created.")
```

## Save data to files

### Subtask:
Save each DataFrame in the `stock_dataframes` dictionary to a separate CSV file, named after the stock symbol.


**Reasoning**:
Iterate through the stock_dataframes dictionary and save each DataFrame to a CSV file with the symbol as the filename.



```python
for symbol, df in stock_dataframes.items():
    filename = f"{symbol}.csv"
    df.to_csv(filename)
    print(f"Saved {filename}")
```
